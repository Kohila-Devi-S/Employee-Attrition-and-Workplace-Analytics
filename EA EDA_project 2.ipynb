## Dataset Cleaning 

### 1) The Overall Attrition Rate

#### Objective: Calculate the percentage of employees who left and stayed in the organization.

```python
df['Attrition'].value_counts()

# Calculate percentage
(df['Attrition'].value_counts(normalize=True) * 100)
```

```python
df['Attrition'].value_counts().plot(
    kind='pie',
    autopct='%1.1f%%',
    figsize=(5,5)
)

plt.title('Overall Employee Attrition Rate')
plt.ylabel('')
plt.show()
```

* The organization experienced an overall employee attrition rate of 16.12%, indicating that approximately 1 in 6 employees left.

### 2) Department with High Attrition Count

#### Objective: Identify the departments with higher employee attrition.

```python
df.groupby('Department')['Attrition'].value_counts()
```

```python
dept = pd.crosstab(
    df['Department'],
    df['Attrition'],
    normalize='index'
) * 100

dept
```

```python
dept['Yes'].plot(
    kind='bar',
    figsize=(7,4)
)

plt.title('Department-wise Attrition Rate')
plt.xlabel('Department')
plt.ylabel('Attrition Percentage')
plt.xticks(rotation=0)
plt.show()
```

* Sales has the highest attrition rate, followed by Human Resources. Research & Development has the lowest attrition rate.

## 10)  Observed Differences between employees who left and those who stayed

#### Objective: Compare employee characteristics across both attrition groups to identify broader patterns.

```python
# A. Numerical variable comparison

df.groupby('Attrition').mean(numeric_only=True)
```

```python
df.groupby('Attrition')[[
    'Age',
    'MonthlyIncome',
    'DistanceFromHome',
    'YearsAtCompany',
    'TotalWorkingYears',
    'YearsInCurrentRole',
    'YearsWithCurrManager'
]].mean()
```

```python
# B. Categorical variable comparison

pd.crosstab(df['MaritalStatus'], df['Attrition'])
```

```python
pd.crosstab(df['BusinessTravel'], df['Attrition'])
```

```python
pd.crosstab(df['EducationField'], df['Attrition'])
```

```python
sns.scatterplot(
    data=df,
    x='Age',
    y='YearsAtCompany',
    hue='Attrition'
)

plt.title('Age vs Years at Company')
plt.show()
```

```python
sns.scatterplot(
    data=df,
    x='TotalWorkingYears',
    y='MonthlyIncome',
    hue='Attrition'
)

plt.title('Monthly Income vs Total Working Years')
plt.show()
```

```python
corr_df = df.select_dtypes(include='number').copy()

corr_df['Attrition'] = df['Attrition'].map({
    'No': 0,
    'Yes': 1
})

corr = corr_df.corr()

plt.figure(figsize=(16,12))
sns.heatmap(corr, annot=True, cmap='coolwarm', fmt='.2f')
plt.title('Correlation Heatmap')
plt.show()
```

```python
sns.violinplot(
    data=df,
    x='Attrition',
    y='YearsAtCompany'
)

plt.title('Years at Company vs Attrition')
plt.show()
```

* Employees who left were generally younger and had lower monthly incomes.

* Attrition showed negative associations with age, monthly income, total working years, and years at the company.

* Employees with longer organizational and job tenure were more likely to stay.

* Overall, lower income and shorter work experience were associated with higher employee attrition.
